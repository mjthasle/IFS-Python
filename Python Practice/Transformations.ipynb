```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
from matplotlib import transforms
from matplotlib.transforms import Affine2D

fig, ax = plt.subplots()
ax.set_xlim([-1, 1])
ax.set_ylim([-1, 1])
triangle_vertices = np.array([[0.3, 0.6], [0.2, 0.2], [0.8, 0.2]])
triangle = Polygon(triangle_vertices)
ax.add_patch(triangle)

#print(Polygon(np.array([[0, 1], [0, 0], [1, 0]])))

#plt.show()

triangle.set_facecolor('k')

plt.rcParams['figure.figsize'] = (10, 10)
```

```python
iden = Affine2D.identity()
#print(iden)

#Affine2D.translate(iden, 0, 1)
f = Affine2D.rotate(iden, np.pi/4)
f.transform(np.array([1, 1]))
d = f.transform(triangle_vertices)

fig, ax = plt.subplots()
ax.set_xlim([-1, 1])
ax.set_ylim([-1, 1])
new_triangle = Polygon(d)
new_triangle.set_facecolor('k')
ax.add_patch(new_triangle)
plt.show()
```
