Given two positive integers $n$ and $k$, codelist returns the list of integers from $0$ to $k^n - 1$ (inclusive) in binary form, as strings, where $k$ is the number of functions in the IFS. In the case of the Cantor set, $k=2$.

```python
def codelist(n,k):
    m = 0
    fns = np.zeros((k**n, n))
    while m < k**n:
        fns.append(bin(m)[2:])
        m = m + 1
    return(fns)
```

Normalize takes a list of strings and makes each string the same length by adding zeros on the left of the short ones.

```python
def normalize(x):
    m = 0
    while m <= len(x) - 1:
        if len(x[m]) < len(max(x)):
            while len(x[m]) < len(max(x)):
                x[m] = '0' + x[m]
        else:
            x[m] = x[m]
        m = m + 1
    return x
```
