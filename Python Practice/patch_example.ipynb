# Rotation example with mathplotlib.patches

Source: https://stackoverflow.com/questions/4285103/matplotlib-rotating-a-patch

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import matplotlib as mpl

fig, ax = plt.subplots()
ax.set_xlim([-1, 1])
ax.set_ylim([-1, 1])

triangle_vertices = np.array([[0.3, 0.6], [0.2, 0.2], [0.8, 0.2]])
triangle = patches.Polygon(triangle_vertices)
triangle2 = patches.Polygon(triangle_vertices)

t2 = mpl.transforms.Affine2D().rotate_deg(-45) + ax.transData # define the transform
triangle2.set_transform(t2)

ax.add_patch(triangle)
ax.add_patch(triangle2)

plt.grid(True)

plt.show()
```
