# Affine2D Examples

Matplotlib includes a framework called `transforms` for arbitrary geometric transformations that is used determine the final position of all elements drawn on the canvas. Within this framework is a class called `Affine2D` that can be used to apply affine transformations to objects on the canvas.

In the cell below, we take a square with vertices at $(3\pm2,3\pm2)$, make two copies, one red and one blue, and apply an affine transformation `t` to the blue one. We explicitly write out the entries of the matrix for the transformation.

```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.set_xlim(-16, 16)
ax.set_ylim(-8, 8)
ax.grid(visible = True, linestyle = '--')
mpl.rcParams['figure.figsize'] = (16, 8)

square1 = mpl.patches.Polygon(np.array([[1, 1], [1, 5], [5, 5], [5, 1]]), edgecolor = 'k', facecolor = 'r', alpha = 0.5)
square2 = mpl.patches.Polygon(np.array([[1, 1], [1, 5], [5, 5], [5, 1]]), edgecolor = 'k', facecolor = 'b', alpha = 0.5)

t = mpl.transforms.Affine2D(np.array([[0.9, 0.25, -7],
                                      [0.5, 1, -2],
                                      [0, 0, 1]])) + ax.transData
square2.set_transform(t)

ax.add_patch(square1)
ax.add_patch(square2)
plt.show()
```

## Built-in affine transformations

`Affine2D` has a lot of built-in methods for common affine transformations, so there is often no need to write out the entire matrix. Below are some of them.

`rotate()`, `rotate_around()`, `rotate_deg()`, `rotate_deg_around()`, `scale()`, `translate()`.

Below, we rotate the blue square counter-clockwise around the origin by an angle of $\frac\pi4$.

```python
fig, ax = plt.subplots()
ax.set_xlim(-16, 16)
ax.set_ylim(-8, 8)
ax.grid(visible = True, linestyle = '--')
mpl.rcParams['figure.figsize'] = (16, 8)

square1 = mpl.patches.Polygon(np.array([[1, 1], [1, 5], [5, 5], [5, 1]]), edgecolor = 'k', facecolor = 'r', alpha = 0.5)
square2 = mpl.patches.Polygon(np.array([[1, 1], [1, 5], [5, 5], [5, 1]]), edgecolor = 'k', facecolor = 'b', alpha = 0.5)

t = mpl.transforms.Affine2D().rotate(np.pi/4) + ax.transData
square2.set_transform(t)

ax.add_patch(square1)
ax.add_patch(square2)
plt.show()
```

Next, we rotate counter-clockwise around the point $(6,0)$ by an angle of $\frac\pi4$.

```python
fig, ax = plt.subplots()
ax.set_xlim(-16, 16)
ax.set_ylim(-8, 8)
ax.grid(visible = True, linestyle = '--')
mpl.rcParams['figure.figsize'] = (16, 8)

square1 = mpl.patches.Polygon(np.array([[1, 1], [1, 5], [5, 5], [5, 1]]), edgecolor = 'k', facecolor = 'r', alpha = 0.5)
square2 = mpl.patches.Polygon(np.array([[1, 1], [1, 5], [5, 5], [5, 1]]), edgecolor = 'k', facecolor = 'b', alpha = 0.5)

t = mpl.transforms.Affine2D().rotate_around(6, 0, np.pi/4) + ax.transData
square2.set_transform(t)

ax.add_patch(square1)
ax.add_patch(square2)
plt.show()
```

`rotate_deg` and `rotate_deg_around` are similar, but the angles are measured in degrees as opposed to radians.

Next, let's scale by a factor of $0.5$.

```python
fig, ax = plt.subplots()
ax.set_xlim(-16, 16)
ax.set_ylim(-8, 8)
ax.grid(visible = True, linestyle = '--')
mpl.rcParams['figure.figsize'] = (16, 8)

square1 = mpl.patches.Polygon(np.array([[1, 1], [1, 5], [5, 5], [5, 1]]), edgecolor = 'k', facecolor = 'r', alpha = 0.5)
square2 = mpl.patches.Polygon(np.array([[1, 1], [1, 5], [5, 5], [5, 1]]), edgecolor = 'k', facecolor = 'b', alpha = 0.5)

t = mpl.transforms.Affine2D().scale(0.5) + ax.transData
square2.set_transform(t)

ax.add_patch(square1)
ax.add_patch(square2)
plt.show()
```

Next, let's translate by the point $(9,-5)$.

```python
fig, ax = plt.subplots()
ax.set_xlim(-16, 16)
ax.set_ylim(-8, 8)
ax.grid(visible = True, linestyle = '--')
mpl.rcParams['figure.figsize'] = (16, 8)

square1 = mpl.patches.Polygon(np.array([[1, 1], [1, 5], [5, 5], [5, 1]]), edgecolor = 'k', facecolor = 'r', alpha = 0.5)
square2 = mpl.patches.Polygon(np.array([[1, 1], [1, 5], [5, 5], [5, 1]]), edgecolor = 'k', facecolor = 'b', alpha = 0.5)

t = mpl.transforms.Affine2D().translate(9, -5) + ax.transData
square2.set_transform(t)

ax.add_patch(square1)
ax.add_patch(square2)
plt.show()
```

Finally, we can compose transformations using '+'. Let's scale by $1/3$, then rotate $30^\circ$ around the point $(5,0)$, then translate by the point $(-5, -4)$.

```python
fig, ax = plt.subplots()
ax.set_xlim(-16, 16)
ax.set_ylim(-8, 8)
ax.grid(visible = True, linestyle = '--')
mpl.rcParams['figure.figsize'] = (16, 8)

square1 = mpl.patches.Polygon(np.array([[1, 1], [1, 5], [5, 5], [5, 1]]), edgecolor = 'k', facecolor = 'r', alpha = 0.5)
square2 = mpl.patches.Polygon(np.array([[1, 1], [1, 5], [5, 5], [5, 1]]), edgecolor = 'k', facecolor = 'b', alpha = 0.5)

t0 = mpl.transforms.Affine2D().scale(1/3)
t1 = mpl.transforms.Affine2D().rotate_deg_around(5, 0, -30)
t2 = mpl.transforms.Affine2D().translate(-5, -4)
t = t0 + t1 + t2 + ax.transData
square2.set_transform(t)

ax.add_patch(square1)
ax.add_patch(square2)
plt.show()
```
